```python
import time

# 1. Création du décorateur
def mesurer_temps(fonction_originale):
    def enveloppe(*args, **kwargs):
        debut = time.time()
        resultat = fonction_originale(*args, **kwargs)  # Exécution de la vraie fonction
        fin = time.time()
        print(f"Temps d'exécution : {fin - debut:.4f} secondes")
        return resultat
    return enveloppe

# 2. Application avec le symbole @
@mesurer_temps
def calcul_lourd():
    sum(i ** 2 for i in range(1_000_000))
    print(f"Temps d'exécution : 2 secondes")

# Quand on appelle calcul_lourd(), c'est en réalité 'enveloppe' qui s'exécute
calcul_lourd()
# Affiche : Temps d'exécution : 0.0642 secondes
```

3. **Les décorateurs (@decorateur)**
Un **décorateur** est une fonction qui enveloppe une autre fonction pour lui ajouter un comportement avant et/ou après son exécution, sans modifier le code interne de cette fonction.

*En Python, les fonctions sont des objets comme les autres : on peut les passer en paramètre et les renvoyer.*
